```python
import numpy as np
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset,DataLoader
from torch.optim import Adam
```

```python
N=1000
X=np.random.randint(low=1,high=10,size=(N,16))
y=np.random.randint(low=0,high=2,size=(N,1))
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
device=torch.device("cuda" if torch.cuda.is_available() else"cpu")
```

```python
print(device)
```

```python
class CustomDataset(Dataset):
  def __init__(self,X,y):
    self.X=X
    self.y=y
  def __len__(self):
    return len(self.X)
  def __getitem__(self, index):
    _,features=self.X.shape
    x = torch.tensor(self.X[index], dtype=torch.float32)
    y = torch.tensor(self.y[index])
    mean=0.0
    std=0.1
    noise=torch.randn_like(torch.tensor(x,dtype=torch.float32))*std+mean

    return x+noise,y
```

```python
train_ds=CustomDataset(X_train,y_train)
test_ds=CustomDataset(X_test,y_test)
train_loader=DataLoader(train_ds,batch_size=8,shuffle=True)
test_loader=DataLoader(test_ds,batch_size=4)
```

```python
for batch_feature,batch_label in iter(train_loader):
  batch_feature=torch.tensor(batch_feature,dtype=torch.float32)
  batch_label=torch.tensor(batch_label,dtype=torch.float32)
  print(batch_feature.shape)
  print(batch_label.shape)
```

```python
class NeuralNetwork(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear1=nn.Linear(16,32)
    self.act1=nn.ReLU()
    self.norm1=nn.BatchNorm1d(32)
    self.drop1=nn.Dropout(p=0.5)
    self.linear2=nn.Linear(32,16)
    self.act2=nn.ReLU()
    self.linear3=nn.Linear(16,1)
    self.act3=nn.Sigmoid()
  def forward(self,x):
    x=self.norm1(self.act1(self.linear1(x)))
    x=self.drop1(x)
    x=self.act2(self.linear2(x))
    x=self.act3(self.linear3(x))
    return x
```

```python
model=NeuralNetwork().to(device)
learning_rate=0.01
criterion=nn.BCELoss()
optimizer=Adam(model.parameters(),lr=learning_rate)
epochs=50
for epoch in range(epochs):
  for batch_feature,batch_label in iter(train_loader):
    batch_feature=torch.tensor(batch_feature,dtype=torch.float32).to(device)
    batch_label=torch.tensor(batch_label,dtype=torch.float32).to(device)
    output=model(batch_feature)
    loss=criterion(output,batch_label)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
  print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")


```

```python
model.eval()
correct=0
threshold=0.5
tot=0
with torch.no_grad():
  for batch_feature,batch_label in iter(test_loader):
    batch_feature=torch.tensor(batch_feature,dtype=torch.float32).to(device)
    batch_label=torch.tensor(batch_label,dtype=torch.float32).to(device)
    output=model(batch_feature)
    preds = (output > threshold).float()
    correct+=(preds==batch_label).sum().item()
    tot+=batch_label.shape[0]
print(correct/tot)


```
